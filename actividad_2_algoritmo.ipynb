#**Ejercicio #1**

```python
while True:
    n = 0
    while n <= 0:
        try:
            n = int(input("Ingrese un número límite N: "))
            if n <= 0:
                print("Dato inválido, debe ser mayor que 0.")
        except ValueError:
            print("Dato inválido, ingrese un número entero.")

    print("Números del 1 al", n, "(con for):")
    for i in range(1, n + 1):
        print(i)

    print("Números pares del 1 al", n, "(con while):")
    contador = 2
    while contador <= n:
        print(contador)
        contador += 2

    continuar = input("¿Desea continuar? (s/n): ").strip().lower()
    if continuar != 's':
        break
```

#**Ejercicio #2**

```python
while True:
  n = 0
  while n <= 0:
    try:
      n = int(input("¿Cuántos números va a ingresar? "))
      if n <= 0:
        print("Dato inválido, debe ser mayor que 0.")
    except ValueError:
      print("Dato inválido, ingrese un número entero.")

  suma = 0
  for i in range(n):
    valido = False
    while not valido:
      try:
        numero = float(input("Ingrese el número " + str(i + 1) + ": "))
        suma += numero
        valido = True
      except ValueError:
        print("Dato inválido, ingrese un valor numérico.")

  promedio = suma / n
  print("Suma:", suma)
  print("Promedio:", promedio)

  continuar = input("¿Desea continuar? (s/n): ").strip().lower()
  if continuar != "s":
    break
```

#**Ejercicio #3**

```python
CLAVE_CORRECTA = "1234"
MAX_INTENTOS = 5

while True:
  intentos = 0
  acceso = False

  while intentos < MAX_INTENTOS and not acceso:
    clave = input("Ingrese la clave: ")

    while clave == "":
      print("la clave no puede estar vacia")
      clave = input("ingrese la clave: ")

    if clave == CLAVE_CORRECTA:
      acceso = True
      print("acceso permitido")
    else:
      intentos += 1
      print("clave incorrecta")

  if not acceso:
    print("acceso bloqueado")

  continuar = input("¿Desea continuar? (s/n): ")
  if continuar.lower() != "s":
    break
```
